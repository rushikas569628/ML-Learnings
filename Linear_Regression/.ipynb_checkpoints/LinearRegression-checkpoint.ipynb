```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression(n_jobs=-1)
```

```python
regression.fit(X_train,y_train)
```

- Here are the weights (inputs) and their corresponding heights (correct answers). Find the relationship between them so you can predict height for new weights.

- - X_train → Weight values (standardized)
- y_train → Actual height values (in inches)
The model tries to learn a line:
\[
\hat y = \theta_0 + \theta_1 x
\]

- \(\theta_0\) → intercept (where the line crosses the y-axis)
- \(\theta_1\) → slope (how predicted height changes as weight changes)

- It calculates how far its predicted heights are from the actual heights. These differences are called residuals (errors).

- model's goal is to find coefficients that minimize the sum of squared residuals.

- \[
J(\theta_0,\theta_1)=\frac{1}{2m}\sum_{i=1}^{m}(\hat y_i-y_i)^2
\] fit() optimize-MSE-costfunction
