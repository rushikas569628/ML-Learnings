##### To find Numerical & Categorical Features

```python
numeric_features=[feature for feature in df_copy.columns if df_copy[feature].dtype!='O']
numeric_features
```

```python
categorical_features=[feature for feature in df_copy.columns if df_copy[feature].dtype=='O']
categorical_features
```

```python
df_copy['App'].dtype
```

```python
numeric_features = df_copy.select_dtypes(include='number').columns.tolist()

categorical_features = df_copy.select_dtypes(exclude='number').columns.tolist()
```

```python
numeric_features
```

```python
categorical_features
```

```python
#proprtion of count data on categorical columns
for col in categorical_features:
    print(df[col].value_counts(normalize=True)*100)
```

```python
plt.figure(figsize=(15,15))

plt.suptitle(
    "Univariate analysis of Numerical features",
    fontsize=20,
    fontweight='bold',
    alpha=0.8,
    y=1.0
)

for i in range(len(numeric_features)):
    plt.subplot(5, 3, i+1)

    sns.kdeplot(#kernel density estimator
        x=df_copy[numeric_features[i]],
        fill=True,
        color='r'
    )

    plt.xlabel(numeric_features[i])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,12))

plt.suptitle(
    "Univariate Analysis of Categorical Features",
    fontsize=20,
    fontweight='bold',
    alpha=0.8,
    y=1.0
)

for i in range(len(categorical_features)):

    plt.subplot(3, 3, i+1)

    sns.countplot(
        x=df_copy[categorical_features[i]],
        palette="Set2"
    )

    plt.xlabel(categorical_features[i])
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
df_copy['Category'].value_counts().plot.pie(
    figsize=(10,10),
    autopct='%1.1f%%'
)

plt.ylabel("")
plt.show()
```

```python
#Top 10 app Categories
category = (
    df_copy["Category"]
    .value_counts()
    .reset_index()
)

category.columns = ["Category", "Count"]

category.head(10)
```

```python
#Top 10 apps barchart
plt.figure(figsize=(15,6))

sns.barplot(
    data=category.head(10),
    x="Category",
    y="Count",
    palette="hls"
)

plt.title("Top 10 App Categories")
plt.xticks(rotation=90)
plt.show()
```
